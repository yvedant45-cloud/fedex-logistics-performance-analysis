# **Project Summary -**

The **FedEx Logistics Performance Analysis** project focuses on exploring logistics and supply chain operations through Exploratory Data Analysis (EDA). The primary objective was to examine shipment patterns, transportation methods, delivery performance, freight costs and vendor-related activities to identify meaningful trends and generate actionable business insights. The analysis was conducted using the SCMS Delivery History dataset provided for the project, within the FedEx logistics business context.

The dataset consisted of **10,324 records and 33 columns**, containing information related to shipment details, countries, vendors, product groups, shipment modes, delivery schedules, freight costs, insurance and product quantities. Important variables included Country, Shipment Mode, Vendor INCO Term, Scheduled Delivery Date, Delivered to Client Date, Product Group, Line Item Quantity, Line Item Value, Weight (Kilograms) and Freight Cost (USD). These variables provided an opportunity to investigate different aspects of logistics operations and understand their relationships.

The project began with data understanding and preparation. The dataset was examined using its dimensions, column information, descriptive statistics, unique values, duplicate records and missing-value distribution. Data cleaning involved creating a separate working copy, removing unnecessary spaces from text fields, checking duplicate records, handling missing categorical values and converting appropriate date and numerical columns into their respective data types. Missing values in fields such as weight, freight cost and insurance were retained where appropriate rather than automatically replaced with zero, as missing information does not necessarily represent an actual zero value. These steps helped improve data consistency and prepare the dataset for further analysis.

Exploratory Data Analysis was performed using **Univariate, Bivariate and Multivariate Analysis**. A total of 20 visualizations were developed using Matplotlib and Seaborn, supported by Pandas and NumPy. Univariate analysis examined shipment mode distribution, country-wise shipment volumes, product groups, vendors, INCO terms and freight cost distribution. Bivariate analysis explored relationships between shipment quantity, product value, shipment weight, freight costs, transportation modes and delivery performance. Multivariate analysis used correlation heatmaps, pair plots and delivery-status comparisons to understand relationships among multiple variables.

The analysis revealed several important findings. Air was the most frequently used shipment mode, followed by Truck. ARV represented the largest product group, while shipment volumes were concentrated among certain countries and vendors. Freight costs showed a highly right-skewed distribution, indicating that most recorded shipments had relatively lower freight costs, while a smaller number involved substantially higher expenses. Correlation analysis identified strong positive relationships between line item quantity and value, as well as between line item value and insurance cost. However, weight and freight cost did not demonstrate a clear linear relationship. Delivery performance analysis also revealed variations across shipment modes, countries and time periods, with both early and late deliveries observed.

Based on these findings, the project identified opportunities to improve transportation planning, freight cost monitoring, vendor coordination, delivery scheduling and data quality. Overall, this analysis demonstrates how Python-based exploratory data analysis can transform raw logistics data into meaningful business intelligence and support informed decision-making. The findings provide a foundation for improving operational visibility, cost efficiency and supply chain performance.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Create a copy of the dataset
df_clean = df.copy()

# Remove extra spaces from text columns
df_clean = df_clean.apply(
    lambda x: x.str.strip() if x.dtype == "object" else x)

# Display dataset shape
df_clean.shape
````

````python
# Check duplicate records
df_clean.duplicated().sum()
````

````python
# Check missing values
df_clean.isnull().sum()
````

````python
# Display rows with missing values
df_clean[df_clean.isnull().any(axis=1)].head()
````

````python
# Handle missing values
# In dosage
df_clean['Dosage'] = df_clean['Dosage'].fillna('Unknown')

# In shipment mode
df_clean['Shipment Mode'] = df_clean['Shipment Mode'].fillna('Unknown')
````

````python
# Convert Scheduled Delivery Date into date format
df_clean['Scheduled Delivery Date'] = pd.to_datetime(
    df_clean['Scheduled Delivery Date'], errors='coerce'
)

# Convert Delivered to Client Date into date format
df_clean['Delivered to Client Date'] = pd.to_datetime(
    df_clean['Delivered to Client Date'], errors='coerce'
)
````

````python
# Convert Weight into numeric format
df_clean['Weight (Kilograms)'] = pd.to_numeric(
    df_clean['Weight (Kilograms)'], errors='coerce'
)

# Convert Freight Cost into numeric format
df_clean['Freight Cost (USD)'] = pd.to_numeric(
    df_clean['Freight Cost (USD)'], errors='coerce')
````

````python
# Check missing values in numeric columns
df_clean[['Weight (Kilograms)', 'Freight Cost (USD)']].isnull().sum()
````

````python
# Check the cleaned dataset shape
df_clean.shape
````

````python
# Check data types after cleaning
df_clean.dtypes
````

````python
# Convert remaining columns into date format
df_clean['PQ First Sent to Client Date'] = pd.to_datetime(df_clean['PQ First Sent to Client Date'], errors='coerce')

df_clean['PO Sent to Vendor Date'] = pd.to_datetime(df_clean['PO Sent to Vendor Date'], errors='coerce')

df_clean['Delivery Recorded Date'] = pd.to_datetime(df_clean['Delivery Recorded Date'], errors='coerce')
````

````python
# Check missing values in date columns
df_clean[['PQ First Sent to Client Date',
          'PO Sent to Vendor Date',
          'Scheduled Delivery Date',
          'Delivered to Client Date',
          'Delivery Recorded Date']].isnull().sum()
````

````python
# Chart - 10 visualization code
# Delivery Time Difference by Shipment Mode

# Calculate delivery time difference in days
df_clean['Delivery Time Difference'] = (
    df_clean['Delivered to Client Date'] -
    df_clean['Scheduled Delivery Date']
).dt.days

# Create box plot to compare delivery performance
plt.figure(figsize=(10, 5))

sns.boxplot(
    x='Shipment Mode',
    y='Delivery Time Difference',
    data=df_clean
)

# Add chart title and labels
plt.title('Delivery Time Difference by Shipment Mode')
plt.xlabel('Shipment Mode')
plt.ylabel('Delivery Time Difference (Days)')

# Add zero reference line
plt.axhline(y=0, color='red', linestyle='--')

# Display the chart
plt.tight_layout()
plt.show()
````

Air and Ocean shipments have median values close to zero, suggesting deliveries are generally around their scheduled dates.

Air shipments show a large number of outliers, including both early and delayed deliveries. Truck and Air Charter also contain several unusual delivery-time observations.
